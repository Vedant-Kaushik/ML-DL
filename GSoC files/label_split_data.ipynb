```python
import os
import shutil
import numpy as np
from sklearn.model_selection import train_test_split

# Define Paths
dataset_path = "./prepared_dataset/"
split_path = "./split_dataset/"

# Paths for Real and Fake images
real_path = os.path.join(dataset_path, "real")
fake_path = os.path.join(dataset_path, "fake")

# Create split dataset folders
for category in ["train", "val", "test"]:
    os.makedirs(os.path.join(split_path, category, "real"), exist_ok=True)
    os.makedirs(os.path.join(split_path, category, "fake"), exist_ok=True)

# Get all image files
real_images = [os.path.join(real_path, f) for f in os.listdir(real_path) if f.endswith(('.png', '.jpg', '.jpeg'))]
fake_images = [os.path.join(fake_path, f) for f in os.listdir(fake_path) if f.endswith(('.png', '.jpg', '.jpeg'))]

# Balance dataset (make fake images count same as real images)
real_count = len(real_images)
fake_images = np.random.choice(fake_images, real_count, replace=False).tolist()

# Split Data (80% Train, 10% Val, 10% Test)
train_real, test_real = train_test_split(real_images, test_size=0.2, random_state=42)
train_fake, test_fake = train_test_split(fake_images, test_size=0.2, random_state=42)

val_real, test_real = train_test_split(test_real, test_size=0.5, random_state=42)
val_fake, test_fake = train_test_split(test_fake, test_size=0.5, random_state=42)

# Function to move files
def move_files(file_list, destination_folder):
    for file in file_list:
        shutil.copy(file, os.path.join(destination_folder, os.path.basename(file)))

# Move files into Train, Val, Test folders
move_files(train_real, os.path.join(split_path, "train", "real"))
move_files(train_fake, os.path.join(split_path, "train", "fake"))

move_files(val_real, os.path.join(split_path, "val", "real"))
move_files(val_fake, os.path.join(split_path, "val", "fake"))

move_files(test_real, os.path.join(split_path, "test", "real"))
move_files(test_fake, os.path.join(split_path, "test", "fake"))

print(f"Dataset Split Complete \nTrain: {len(train_real)} real, {len(train_fake)} fake\nVal: {len(val_real)} real, {len(val_fake)} fake\nTest: {len(test_real)} real, {len(test_fake)} fake")
```
