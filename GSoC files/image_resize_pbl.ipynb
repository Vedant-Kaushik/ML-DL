```python
import os
import cv2

# Define the base path for images
base_path = "./train_sample_images/"  # Assuming images are stored here

# Ensure output directory exists
output_path = "./processed_images/"
os.makedirs(output_path, exist_ok=True)

# Function to get filename without extension
def get_filename_only(file_path):
    return os.path.splitext(os.path.basename(file_path))[0]

# Processing each image in the folder
for filename in os.listdir(base_path):
    file_path = os.path.join(base_path, filename)

    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):  # Process only images
        print(f"Processing {filename}...")

        # Read the image
        image = cv2.imread(file_path)

        if image is None:
            print(f"Skipping {filename}, unable to read.")
            continue

        # Get original dimensions
        height, width, _ = image.shape
        print(f"Original Dimensions: {width}x{height}")

        # Resize based on width
        if width < 300:
            scale_ratio = 2
        elif width > 1900:
            scale_ratio = 0.33
        elif width > 1000:
            scale_ratio = 0.5
        else:
            scale_ratio = 1

        print(f"Scale Ratio: {scale_ratio}")

        # Resize the image
        new_width = int(width * scale_ratio)
        new_height = int(height * scale_ratio)
        resized_image = cv2.resize(image, (new_width, new_height), interpolation=cv2.INTER_AREA)

        print(f"Resized Dimensions: {new_width}x{new_height}")

        # Save the processed image
        new_filename = f"{get_filename_only(filename)}_processed.png"
        output_filepath = os.path.join(output_path, new_filename)
        cv2.imwrite(output_filepath, resized_image)
        print(f"Saved: {output_filepath}")

print("Processing complete!")
```
